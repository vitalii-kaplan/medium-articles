```python
################################################################################################################################################################
## X_Partitioner # `68`
# Node state: `EXECUTED`
# Input: [Port 1] 90:1 from CSV Reader #90
# Output: [Port 1] 68:1 to 14 Equal Size Sampling #14; [Port 2] 68:2 to 19 Random Forest Predictor #19
# https://hub.knime.com/knime/extensions/org.knime.features.base/latest/org.knime.base.node.meta.xvalidation.XValidatePartitionerFactory
__xpart_src_68 = context['90:1']  # input table (frozen for the whole CV loop)
# Plan cross-validation folds and open the loop
_leave_one_out = False
_k_req = 10
_shuffle = False
_seed = None
_stratified = True
_class_col = 'Churn (Y/N)'

n_samples = int(len(__xpart_src_68))

_y = None
if _stratified and _class_col and _class_col in __xpart_src_68.columns:
    _y = __xpart_src_68[_class_col].astype('object').where(pd.notna(__xpart_src_68[_class_col]), '__NA__')

folds = []  # list of (train_idx, test_idx)

if _leave_one_out:
    if n_samples >= 2:
        try:
            splitter = LeaveOneOut()
            for tr, te in splitter.split(__xpart_src_68, _y if _y is not None else None):
                folds.append((tr, te))
        except Exception:
            folds = []
    # else: too few samples → leave folds empty (fallback below)
else:
    # KFold / StratifiedKFold branch
    if n_samples < 2:
        folds = []  # fallback below will handle
    else:
        _k_eff = int(_k_req) if _k_req is not None else 2
        _k_eff = max(2, min(_k_eff, n_samples))
        _did_strat = False
        if _stratified and _y is not None:
            try:
                # Ensure each class has at least _k_eff members for stratification
                vc = pd.Series(_y).value_counts(dropna=False)
                if (vc.min() >= _k_eff) and (len(vc) >= 2):
                    splitter = StratifiedKFold(n_splits=_k_eff, shuffle=_shuffle, random_state=_seed if _shuffle else None)
                    for tr, te in splitter.split(__xpart_src_68, _y):
                        folds.append((tr, te))
                    _did_strat = True
            except Exception:
                folds = []
        if not _did_strat:
            try:
                splitter = KFold(n_splits=_k_eff, shuffle=_shuffle, random_state=_seed if _shuffle else None)
                for tr, te in splitter.split(__xpart_src_68):
                    folds.append((tr, te))
            except Exception:
                folds = []

# Fallback: if splitting produced no folds, emit a single all-train / empty-test fold
if not folds:
    folds = [(np.arange(len(__xpart_src_68)), np.array([], dtype=int))]

loop_state = {
    'node_id': '68',
    'k': len(folds),
    'current': 0,
    'classColumn': _class_col,
    'stratified': bool(_y is not None),
    'shuffle': _shuffle,
    'seed': _seed,
    'folds': folds,
}
context['__loop__:68'] = loop_state
context['__flowvar__:68:maxIterations'] = len(folds)
context.pop('__xagg__:68:accum', None)
context.pop('__xagg__:68:is_complete', None)

for __fold_idx, (__tr, __te) in enumerate(folds):
    context['__flowvar__:68:currentIteration'] = int(__fold_idx)
    context['__loop__:68']['current'] = int(__fold_idx)
    train_df = __xpart_src_68.iloc[__tr].reset_index(drop=True)
    test_df  = __xpart_src_68.iloc[__te].reset_index(drop=True)
    context['68:1'] = train_df
    context['68:2'] = test_df
    ################################################################################################################################################################
    ## Equal Size Sampling # `14`
    # Node state: `EXECUTED`
    # Input: [Port 1] 68:1 from X_Partitioner #68
    # Output: [Port 1] 14:1 to 16 Random Forest Learner #16
    # https://hub.knime.com/knime/extensions/org.knime.features.base/latest/org.knime.base.node.preproc.equalsizesampling.EqualSizeSamplingNodeFactory
    df = context['68:1']  # input table
    out_df = df.copy()
    _class = 'Churn (Y/N)'
    _seed = 1
    if df.empty:
        out_df = df.iloc[0:0]
    else:
        groups = [g for _, g in df.groupby(_class, dropna=False, sort=False)]
        if not groups:
            out_df = df.iloc[0:0]
        else:
            min_count = min(len(g) for g in groups)
            if min_count <= 0:
                out_df = df.iloc[0:0]
            else:
                parts = [resample(g, replace=False, n_samples=min_count, random_state=_seed) for g in groups]
                out_df = pd.concat(parts, axis=0).sort_index() if parts else df.iloc[0:0]
    context['14:1'] = out_df
    ################################################################################################################################################################
    ## Random Forest Learner # `16`
    # Node state: `EXECUTED`
    # Input: [Port 1] 14:1 from Equal Size Sampling #14
    # Output: [Port 3] 16:3 to 19 Random Forest Predictor #19
    # https://hub.knime.com/knime/extensions/org.knime.features.ensembles/latest/org.knime.base.node.mine.treeensemble2.node.randomforest.learner.classification.RandomForestClassificationLearnerNodeFactory2
    df = context['14:1']  # input table
    out_df = df.copy()
    _target = 'Churn (Y/N)'
    _include = ['Address', 'Age', 'Callcard', 'Callid', 'Callwait', 'Cardmon', 'Cardten', 'Confer', 'Custcat', 'Ebill', 'Education', 'Employ', 'Equip', 'Equipmon', 'Equipten', 'Forward', 'GeFemaleder', 'Income', 'Internet', 'lnInc', 'LogLong', 'LongMon', 'LongTen', 'Marital', 'Multline', 'Pager', 'Region', 'Reside', 'Retire', 'Tenure', 'Tollfree', 'Tollmon', 'Tollten', 'Voice', 'Wireless', 'Wiremon', 'Wireten']
    feat_cols = [c for c in _include if c in df.columns]
    X = df[feat_cols].copy()
    y = df[_target].copy()
    _n_estimators = int(100)
    _max_depth = None
    _min_samples_split = int(2)
    _min_samples_leaf = int(1)
    _bootstrap = True
    _max_samples = None
    _max_features = 'sqrt'  # mapped from KNIME columnSamplingMode
    _seed = int(1)  # from KNIME settings (defaulted to 1 if missing)
    # KNIME splitCriterion='InformationGainRatio' (sklearn RF supports 'gini'/'entropy' internally)
    # KNIME missingValueHandling='XGBoost' (not supported in sklearn RF; impute beforehand)
    # KNIME useAverageSplitPoints=True (not applicable in sklearn RF)
    # KNIME useBinaryNominalSplits=True (not applicable in sklearn RF)
    model = RandomForestClassifier(
        n_estimators=_n_estimators,
        max_depth=_max_depth,
        min_samples_split=_min_samples_split,
        min_samples_leaf=_min_samples_leaf,
        bootstrap=_bootstrap,
        max_samples=_max_samples,
        max_features=_max_features,
        random_state=_seed,
        n_jobs=-1,
    )
    model.fit(X, y)
    fi = getattr(model, 'feature_importances_', None)
    if fi is not None:
        importances_df = pd.DataFrame({'feature': feat_cols, 'importance': fi})
        importances_df = importances_df.sort_values('importance', ascending=False, kind='mergesort').reset_index(drop=True)
    else:
        importances_df = pd.DataFrame(columns=['feature','importance'])
    summary_df = pd.DataFrame([{'n_estimators': _n_estimators, 'max_depth': _max_depth, 'min_samples_split': _min_samples_split, 'min_samples_leaf': _min_samples_leaf, 'bootstrap': _bootstrap, 'max_samples': _max_samples, 'max_features': _max_features, 'n_features': len(feat_cols), 'classes': ','.join(map(str, getattr(model, 'classes_', [])))}])
    bundle = {
        'estimator': model,
        'model': model,  # alias
        'features': list(feat_cols),
        'feature_cols': list(feat_cols),
        'target': _target,
        'classes': list(getattr(model, 'classes_', [])),
        'n_estimators': _n_estimators,
        'max_depth': _max_depth,
        'min_samples_split': _min_samples_split,
        'min_samples_leaf': _min_samples_leaf,
        'bootstrap': _bootstrap,
        'max_samples': _max_samples,
        'max_features': _max_features,
        'random_state': _seed,
    }
    context['16:3'] = bundle
    ################################################################################################################################################################
    ## Random Forest Predictor # `19`
    # Node state: `EXECUTED`
    # Input: [Port 1] 16:3 from Random Forest Learner #16; [Port 2] 68:2 from X_Partitioner #68
    # Output: [Port 1] 19:1 to 63 X_Aggregator #63
    # https://hub.knime.com/knime/extensions/org.knime.features.ensembles/latest/org.knime.base.node.mine.treeensemble2.node.randomforest.predictor.classification.RandomForestClassificationPredictorNodeFactory2
    model_key = '16:3'
    data_key = '68:2'
    model_obj = context[model_key]
    df = context[data_key]
    out_df = df.copy()

    # Normalize to a dict-like bundle with expected keys
    if isinstance(model_obj, dict):
        bundle = model_obj
    else:
        bundle = {'estimator': model_obj}

    est = bundle.get('estimator') or bundle.get('model')
    feat = bundle.get('features') or getattr(est, 'feature_names_in_', None)
    tgt  = bundle.get('target') or bundle.get('y_col') or bundle.get('target_name')
    classes = list(bundle.get('classes') or getattr(est, 'classes_', []))

    # Normalize features to a plain list
    if isinstance(feat, (list, tuple)):
        feat_list = list(feat)
    elif getattr(feat, 'tolist', None):
        feat_list = list(feat.tolist())
    elif feat is None:
        feat_list = []
    else:
        try:
            feat_list = list(feat)
        except Exception:
            feat_list = []

    # Fallback: use all columns except target, if target known
    if not feat_list:
        if tgt and tgt in out_df.columns:
            feat_list = [c for c in out_df.columns if c != tgt]
        else:
            feat_list = list(out_df.columns)

    missing = [c for c in feat_list if c not in out_df.columns]
    if missing:
        raise KeyError(f'Missing feature columns in data: {missing}')
    X = out_df[feat_list]

    pred_col = f'Prediction ({tgt or "target"})'
    pred = est.predict(X)
    out_df[pred_col] = pd.Series(pred, index=out_df.index).astype('object')

    # Probability matrix (if available)
    proba = None
    if hasattr(est, 'predict_proba'):
        proba = est.predict_proba(X)
        if not classes and getattr(proba, 'shape', (0, 0))[1] == 2:
            classes = ['class0', 'class1']
    _suf = '_RF'
    if proba is not None and classes:
        for j, cls in enumerate(classes):
            cname = f"P ({tgt}={cls}){_suf}"
            out_df[cname] = proba[:, j]
    # Add a prediction confidence column if probabilities are available
    if proba is not None:
        conf = proba.max(axis=1)
        out_df[f"{pred_col} (confidence)"] = conf
    context['19:1'] = out_df
    ################################################################################################################################################################
    ## X_Aggregator # `63`
    # Node state: `EXECUTED`
    # Input: [Port 1] 19:1 from Random Forest Predictor #19
    # Output: [Port 1] 63:1 to 18 Scorer #18; [Port 1] 63:1 to 35 Column Appender #35
    # https://hub.knime.com/knime/extensions/org.knime.features.base/latest/org.knime.base.node.meta.xvalidation.AggregateOutputNodeFactory
    df = context['19:1']  # per-fold input table
    # Aggregate per-fold outputs; publish only on the final fold
    _pred_col = 'Prediction (Churn (Y/N))'
    _targ_col = 'Churn (Y/N)'
    _add_fold = False

    # Bind to an active loop state written by X-Partitioner
    _loop = None
    _loop_id = None
    # Prefer a loop that is not yet complete
    _loop_keys = [k for k in list(context.keys()) if isinstance(k, str) and k.startswith('__loop__:')]
    if _loop_keys:
        # Try to pick a loop whose accumulator is not marked complete
        _cands = []
        for _lk in _loop_keys:
            try:
                _lid = _lk.split(':', 1)[1]
            except Exception:
                _lid = None
            _lobj = context.get(_lk, None)
            _is_done = bool(context.get(f'__xagg__:{_lid}:is_complete', False)) if _lid else False
            if isinstance(_lobj, dict) and not _is_done:
                _cands.append((_lid, _lobj))
        if _cands:
            # If multiple, choose the one with the greatest 'current' (most in-progress)
            _lid, _lobj = sorted(_cands, key=lambda t: int(t[1].get('current', 0)), reverse=True)[0]
            _loop_id, _loop = _lid, _lobj
        else:
            # Fallback: last loop key (lexicographically) if all are complete
            _loop_keys.sort()
            _loop_key = _loop_keys[-1]
            _loop = context.get(_loop_key, None)
            try:
                _loop_id = _loop_key.split(':', 1)[1]
            except Exception:
                _loop_id = str(_loop_key)
    else:
        _loop = None
        _loop_id = str('63')

    _cur = int(_loop.get('current', 0)) if isinstance(_loop, dict) else 0
    _k   = int(_loop.get('k', 1))       if isinstance(_loop, dict) else 1

    # Reset accumulator at the beginning of the loop to prevent stale concatenation
    if int(_cur) == 0:
        context.pop(f'__xagg__:{_loop_id}:accum', None)
        context.pop(f'__xagg__:{_loop_id}:is_complete', None)

    # Current fold input
    cur_df = df.copy()
    if _add_fold:
        cur_df = cur_df.copy()
        cur_df['Fold'] = _cur

    # Optional: sanity-check columns (non-fatal)
    if _pred_col and _pred_col not in cur_df.columns:
        _cand = [c for c in cur_df.columns if c.startswith('Prediction (')]
        if _cand:
            _pred_col = _cand[0]
    # target column is optional for this aggregation

    # Append this fold to an accumulator under the loop id
    _acc_key = f'__xagg__:{_loop_id}:accum'
    _acc = context.get(_acc_key, None)
    if not isinstance(_acc, list):
        _acc = []
    _acc.append(cur_df)
    context[_acc_key] = _acc

    # Mark completion state and publish only on final fold
    context[f'__xagg__:{_loop_id}:is_complete'] = (len(_acc) >= max(int(_k), 1))
    if int(_cur) >= max(int(_k), 1) - 1:
        out_df = pd.concat(_acc, ignore_index=True) if _acc else cur_df.iloc[0:0].copy()
        context['63:1'] = out_df
        # (end of loop) — downstream nodes will run once, outside the loop body
```

```python
################################################################################################################################################################
## X_Partitioner # `71`
# Node state: `EXECUTED`
# Input: [Port 1] 90:1 from CSV Reader #90
# Output: [Port 1] 71:1 to 21 Equal Size Sampling #21; [Port 2] 71:2 to 25 MultiLayerPerceptron Predictor #25
# https://hub.knime.com/knime/extensions/org.knime.features.base/latest/org.knime.base.node.meta.xvalidation.XValidatePartitionerFactory
__xpart_src_71 = context['90:1']  # input table (frozen for the whole CV loop)
# Plan cross-validation folds and open the loop
_leave_one_out = False
_k_req = 10
_shuffle = False
_seed = None
_stratified = True
_class_col = 'Churn (Y/N)'

n_samples = int(len(__xpart_src_71))

_y = None
if _stratified and _class_col and _class_col in __xpart_src_71.columns:
    _y = __xpart_src_71[_class_col].astype('object').where(pd.notna(__xpart_src_71[_class_col]), '__NA__')

folds = []  # list of (train_idx, test_idx)

if _leave_one_out:
    if n_samples >= 2:
        try:
            splitter = LeaveOneOut()
            for tr, te in splitter.split(__xpart_src_71, _y if _y is not None else None):
                folds.append((tr, te))
        except Exception:
            folds = []
    # else: too few samples → leave folds empty (fallback below)
else:
    # KFold / StratifiedKFold branch
    if n_samples < 2:
        folds = []  # fallback below will handle
    else:
        _k_eff = int(_k_req) if _k_req is not None else 2
        _k_eff = max(2, min(_k_eff, n_samples))
        _did_strat = False
        if _stratified and _y is not None:
            try:
                # Ensure each class has at least _k_eff members for stratification
                vc = pd.Series(_y).value_counts(dropna=False)
                if (vc.min() >= _k_eff) and (len(vc) >= 2):
                    splitter = StratifiedKFold(n_splits=_k_eff, shuffle=_shuffle, random_state=_seed if _shuffle else None)
                    for tr, te in splitter.split(__xpart_src_71, _y):
                        folds.append((tr, te))
                    _did_strat = True
            except Exception:
                folds = []
        if not _did_strat:
            try:
                splitter = KFold(n_splits=_k_eff, shuffle=_shuffle, random_state=_seed if _shuffle else None)
                for tr, te in splitter.split(__xpart_src_71):
                    folds.append((tr, te))
            except Exception:
                folds = []

# Fallback: if splitting produced no folds, emit a single all-train / empty-test fold
if not folds:
    folds = [(np.arange(len(__xpart_src_71)), np.array([], dtype=int))]

loop_state = {
    'node_id': '71',
    'k': len(folds),
    'current': 0,
    'classColumn': _class_col,
    'stratified': bool(_y is not None),
    'shuffle': _shuffle,
    'seed': _seed,
    'folds': folds,
}
context['__loop__:71'] = loop_state
context['__flowvar__:71:maxIterations'] = len(folds)
context.pop('__xagg__:71:accum', None)
context.pop('__xagg__:71:is_complete', None)

for __fold_idx, (__tr, __te) in enumerate(folds):
    context['__flowvar__:71:currentIteration'] = int(__fold_idx)
    context['__loop__:71']['current'] = int(__fold_idx)
    train_df = __xpart_src_71.iloc[__tr].reset_index(drop=True)
    test_df  = __xpart_src_71.iloc[__te].reset_index(drop=True)
    context['71:1'] = train_df
    context['71:2'] = test_df
    ################################################################################################################################################################
    ## Equal Size Sampling # `21`
    # Node state: `EXECUTED`
    # Input: [Port 1] 71:1 from X_Partitioner #71
    # Output: [Port 1] 21:1 to 23 RProp MLP Learner #23
    # https://hub.knime.com/knime/extensions/org.knime.features.base/latest/org.knime.base.node.preproc.equalsizesampling.EqualSizeSamplingNodeFactory
    df = context['71:1']  # input table
    out_df = df.copy()
    _class = 'Churn (Y/N)'
    _seed = 1
    if df.empty:
        out_df = df.iloc[0:0]
    else:
        groups = [g for _, g in df.groupby(_class, dropna=False, sort=False)]
        if not groups:
            out_df = df.iloc[0:0]
        else:
            min_count = min(len(g) for g in groups)
            if min_count <= 0:
                out_df = df.iloc[0:0]
            else:
                parts = [resample(g, replace=False, n_samples=min_count, random_state=_seed) for g in groups]
                out_df = pd.concat(parts, axis=0).sort_index() if parts else df.iloc[0:0]
    context['21:1'] = out_df
    ################################################################################################################################################################
    ## RProp MLP Learner # `23`
    # Node state: `EXECUTED`
    # Input: [Port 1] 21:1 from Equal Size Sampling #21
    # Output: [Port 1] 23:1 to 25 MultiLayerPerceptron Predictor #25
    # https://hub.knime.com/knime/extensions/org.knime.features.base/latest/org.knime.base.node.mine.neural.rprop.RPropNodeFactory2
    df = context['21:1']  # input table
    out_df = df.copy()
    _target = 'Churn (Y/N)'
    if _target not in df.columns:
        raise KeyError(f'MLP Learner: target column not found: {_target!r}')
    num_like = df.select_dtypes(include=['number','bool','boolean','Int64','Float64']).columns.tolist()
    feat_cols = [c for c in num_like if c != _target]
    if not feat_cols:
        raise ValueError('MLP Learner: no numeric/bool feature columns found (after excluding target)')
    X = df[feat_cols].copy()
    y = df[_target].copy()
    # ignoremv=False → keep rows; impute inside the pipeline so sklearn can fit
    def _is_binary_col(s):
        try:
            v = pd.to_numeric(s, errors='coerce')
        except Exception:
            return False
        u = pd.unique(v.dropna())
        return set(map(float, u)) <= {0.0, 1.0} and len(u) <= 2

    _bin_cols = [c for c in feat_cols if _is_binary_col(X[c])]
    _cont_cols = [c for c in feat_cols if c not in _bin_cols]
    bin_pipe = Pipeline([('impute', SimpleImputer(strategy='most_frequent'))])
    cont_pipe = Pipeline([('impute', SimpleImputer(strategy='median')), ('scale', StandardScaler())])
    ct = ColumnTransformer([
        ('bin',  bin_pipe, _bin_cols),
        ('cont', cont_pipe, _cont_cols),
    ], remainder='drop', sparse_threshold=0.0)

    _hidden_layer_sizes = tuple([16] * 1)
    _max_iter = int(100)
    _seed = int(1)
    mlp = MLPClassifier(
        hidden_layer_sizes=_hidden_layer_sizes,
        activation='tanh',
        solver='lbfgs',
        max_iter=_max_iter,
        random_state=_seed,
    )
    model = make_pipeline(ct, mlp)
    model.fit(X, y)
    classes = list(getattr(model[-1], 'classes_', []))
    n_iter_ = int(getattr(model[-1], 'n_iter_', 0))
    loss_ = float(getattr(model[-1], 'loss_', float('nan')))
    converged = bool(n_iter_ < _max_iter)  # lbfgs usually converges before max_iter
    summary_df = pd.DataFrame([{'n_features': len(feat_cols),
                                   'classes': ','.join(map(str, classes)),
                                   'hidden_layer_sizes': str(_hidden_layer_sizes),
                                   'max_iter': _max_iter, 'n_iter_': n_iter_,
                                   'loss_': loss_, 'converged': converged}])
    bundle = {
        'estimator': model,
        'model': model,  # alias
        'features': list(feat_cols),
        'feature_cols': list(feat_cols),
        'target': _target,
        'classes': classes,
        'hidden_layer_sizes': _hidden_layer_sizes,
        'max_iter': _max_iter,
        'random_state': _seed,
    }
    context['23:1'] = bundle
    ################################################################################################################################################################
    ## MultiLayerPerceptron Predictor # `25`
    # Node state: `EXECUTED`
    # Input: [Port 1] 23:1 from RProp MLP Learner #23; [Port 2] 71:2 from X_Partitioner #71
    # Output: [Port 1] 25:1 to 64 X_Aggregator #64
    # https://hub.knime.com/knime/extensions/org.knime.features.base/latest/org.knime.base.node.mine.neural.mlp2.MLPPredictorNodeFactory
    model_key = '23:1'
    data_key = '71:2'
    model_obj = context[model_key]
    df = context[data_key]
    out_df = df.copy()

    # Normalize to a dict-like bundle with expected keys
    if isinstance(model_obj, dict):
        bundle = model_obj
    else:
        bundle = {'estimator': model_obj}

    est = bundle.get('estimator') or bundle.get('model')
    feat = bundle.get('features') or getattr(est, 'feature_names_in_', None)
    tgt  = bundle.get('target') or bundle.get('y_col') or bundle.get('target_name')
    classes = list(bundle.get('classes') or getattr(est, 'classes_', []))

    # Normalize features to a plain list
    if isinstance(feat, (list, tuple)):
        feat_list = list(feat)
    elif getattr(feat, 'tolist', None):
        feat_list = list(feat.tolist())
    elif feat is None:
        feat_list = []
    else:
        try:
            feat_list = list(feat)
        except Exception:
            feat_list = []

    # Feature fallback if none provided by bundle/estimator
    if not feat_list:
        if tgt and tgt in out_df.columns:
            feat_list = [c for c in out_df.columns if c != tgt]
        else:
            feat_list = list(out_df.columns)

    missing = [c for c in feat_list if c not in out_df.columns]
    if missing:
        raise KeyError(f'Missing feature columns in data: {missing}')
    X = out_df[feat_list]

    pred_col = f'Prediction ({tgt or "target"})'
    pred = est.predict(X)
    out_df[pred_col] = pd.Series(pred, index=out_df.index).astype('object')

    # Class probability columns (if supported by estimator)
    if hasattr(est, 'predict_proba'):
        proba = est.predict_proba(X)
        if not classes and getattr(proba, 'shape', (0, 0))[1] == 2:
            classes = ['class0', 'class1']
        _suf = '_NN'
        for j, cls in enumerate(classes):
            cname = f"P ({tgt}={cls}){_suf}"
            out_df[cname] = proba[:, j]
    context['25:1'] = out_df
    ################################################################################################################################################################
    ## X_Aggregator # `64`
    # Node state: `EXECUTED`
    # Input: [Port 1] 25:1 from MultiLayerPerceptron Predictor #25
    # Output: [Port 1] 64:1 to 24 Scorer #24; [Port 1] 64:1 to 35 Column Appender #35
    # https://hub.knime.com/knime/extensions/org.knime.features.base/latest/org.knime.base.node.meta.xvalidation.AggregateOutputNodeFactory
    df = context['25:1']  # per-fold input table
    # Aggregate per-fold outputs; publish only on the final fold
    _pred_col = 'Prediction (Churn (Y/N))'
    _targ_col = 'Churn (Y/N)'
    _add_fold = False

    # Bind to an active loop state written by X-Partitioner
    _loop = None
    _loop_id = None
    # Prefer a loop that is not yet complete
    _loop_keys = [k for k in list(context.keys()) if isinstance(k, str) and k.startswith('__loop__:')]
    if _loop_keys:
        # Try to pick a loop whose accumulator is not marked complete
        _cands = []
        for _lk in _loop_keys:
            try:
                _lid = _lk.split(':', 1)[1]
            except Exception:
                _lid = None
            _lobj = context.get(_lk, None)
            _is_done = bool(context.get(f'__xagg__:{_lid}:is_complete', False)) if _lid else False
            if isinstance(_lobj, dict) and not _is_done:
                _cands.append((_lid, _lobj))
        if _cands:
            # If multiple, choose the one with the greatest 'current' (most in-progress)
            _lid, _lobj = sorted(_cands, key=lambda t: int(t[1].get('current', 0)), reverse=True)[0]
            _loop_id, _loop = _lid, _lobj
        else:
            # Fallback: last loop key (lexicographically) if all are complete
            _loop_keys.sort()
            _loop_key = _loop_keys[-1]
            _loop = context.get(_loop_key, None)
            try:
                _loop_id = _loop_key.split(':', 1)[1]
            except Exception:
                _loop_id = str(_loop_key)
    else:
        _loop = None
        _loop_id = str('64')

    _cur = int(_loop.get('current', 0)) if isinstance(_loop, dict) else 0
    _k   = int(_loop.get('k', 1))       if isinstance(_loop, dict) else 1

    # Reset accumulator at the beginning of the loop to prevent stale concatenation
    if int(_cur) == 0:
        context.pop(f'__xagg__:{_loop_id}:accum', None)
        context.pop(f'__xagg__:{_loop_id}:is_complete', None)

    # Current fold input
    cur_df = df.copy()
    if _add_fold:
        cur_df = cur_df.copy()
        cur_df['Fold'] = _cur

    # Optional: sanity-check columns (non-fatal)
    if _pred_col and _pred_col not in cur_df.columns:
        _cand = [c for c in cur_df.columns if c.startswith('Prediction (')]
        if _cand:
            _pred_col = _cand[0]
    # target column is optional for this aggregation

    # Append this fold to an accumulator under the loop id
    _acc_key = f'__xagg__:{_loop_id}:accum'
    _acc = context.get(_acc_key, None)
    if not isinstance(_acc, list):
        _acc = []
    _acc.append(cur_df)
    context[_acc_key] = _acc

    # Mark completion state and publish only on final fold
    context[f'__xagg__:{_loop_id}:is_complete'] = (len(_acc) >= max(int(_k), 1))
    if int(_cur) >= max(int(_k), 1) - 1:
        out_df = pd.concat(_acc, ignore_index=True) if _acc else cur_df.iloc[0:0].copy()
        context['64:1'] = out_df
        # (end of loop) — downstream nodes will run once, outside the loop body
```

```python
################################################################################################################################################################
## Scorer # `24`
# Node state: `EXECUTED`
# Input: [Port 1] 64:1 from X_Aggregator #64
# Output: [Port 1] 24:1 to 86 CSV Writer #86
# https://hub.knime.com/knime/extensions/org.knime.features.base/latest/org.knime.base.node.mine.scorer.accuracy.AccuracyScorer2NodeFactory
df = context['64:1']  # input table with truth & prediction
out_df = df.copy()  # not strictly required, but keeps pattern consistent
_truth_col = 'Churn (Y/N)'
_pred_col  = 'Prediction (Churn (Y/N))'

# Validate required columns exist
missing = []
if _truth_col not in df.columns: missing.append(_truth_col)
if _pred_col not in df.columns:  missing.append(_pred_col)
if missing:
    raise KeyError(f'Scorer: missing required column(s): {missing}')

pair = df[[_truth_col, _pred_col]].copy()
pair = pair.dropna(subset=[_truth_col, _pred_col])  # KNIME: ignore.missing.values = true

y_true = pair[_truth_col].astype('object')
y_pred = pair[_pred_col].astype('object')

# Metrics
acc = float(accuracy_score(y_true, y_pred)) if len(pair) else float('nan')
err = (1.0 - acc) if pd.notna(acc) else float('nan')
correct = int((y_true == y_pred).sum())
false   = int((y_true != y_pred).sum())
try:
    kappa = float(cohen_kappa_score(y_true, y_pred)) if len(pair) else float('nan')
except Exception:
    kappa = float('nan')

# Confusion matrix with a stable label list (union from both columns, order of appearance)
labels = pd.unique(pd.concat([y_true, y_pred], ignore_index=True))
if len(labels) == 0:
    cm_df = pd.DataFrame(columns=pd.Index([], name='Predicted'), index=pd.Index([], name='Actual'))
else:
    cm = confusion_matrix(y_true, y_pred, labels=labels)
    cm_df = pd.DataFrame(cm, index=pd.Index(labels, name='Actual'), columns=pd.Index(labels, name='Predicted'))

summary_rows = [
    {'Metric': 'Accuracy',       'Value': acc},
    {'Metric': 'Error',          'Value': err},
    {'Metric': '#Correct',       'Value': correct},
    {'Metric': '#False',         'Value': false},
    {'Metric': "Cohen's kappa", 'Value': kappa},
]
summary_df = pd.DataFrame(summary_rows)

# Outputs:
# Port 1 → confusion matrix, Port 2 → summary metrics
cm_out = cm_df
stats_out = summary_df
print(cm_df)
print(summary_df)
context['24:1'] = cm_out
context['24:2'] = stats_out
```

```python
################################################################################################################################################################
## X_Partitioner # `69`
# Node state: `EXECUTED`
# Input: [Port 1] 90:1 from CSV Reader #90
# Output: [Port 1] 69:1 to 44 Equal Size Sampling #44; [Port 2] 69:2 to 49 Gradient Boosted Trees Predictor #49
# https://hub.knime.com/knime/extensions/org.knime.features.base/latest/org.knime.base.node.meta.xvalidation.XValidatePartitionerFactory
__xpart_src_69 = context['90:1']  # input table (frozen for the whole CV loop)
# Plan cross-validation folds and open the loop
_leave_one_out = False
_k_req = 10
_shuffle = False
_seed = None
_stratified = True
_class_col = 'Churn (Y/N)'

n_samples = int(len(__xpart_src_69))

_y = None
if _stratified and _class_col and _class_col in __xpart_src_69.columns:
    _y = __xpart_src_69[_class_col].astype('object').where(pd.notna(__xpart_src_69[_class_col]), '__NA__')

folds = []  # list of (train_idx, test_idx)

if _leave_one_out:
    if n_samples >= 2:
        try:
            splitter = LeaveOneOut()
            for tr, te in splitter.split(__xpart_src_69, _y if _y is not None else None):
                folds.append((tr, te))
        except Exception:
            folds = []
    # else: too few samples → leave folds empty (fallback below)
else:
    # KFold / StratifiedKFold branch
    if n_samples < 2:
        folds = []  # fallback below will handle
    else:
        _k_eff = int(_k_req) if _k_req is not None else 2
        _k_eff = max(2, min(_k_eff, n_samples))
        _did_strat = False
        if _stratified and _y is not None:
            try:
                # Ensure each class has at least _k_eff members for stratification
                vc = pd.Series(_y).value_counts(dropna=False)
                if (vc.min() >= _k_eff) and (len(vc) >= 2):
                    splitter = StratifiedKFold(n_splits=_k_eff, shuffle=_shuffle, random_state=_seed if _shuffle else None)
                    for tr, te in splitter.split(__xpart_src_69, _y):
                        folds.append((tr, te))
                    _did_strat = True
            except Exception:
                folds = []
        if not _did_strat:
            try:
                splitter = KFold(n_splits=_k_eff, shuffle=_shuffle, random_state=_seed if _shuffle else None)
                for tr, te in splitter.split(__xpart_src_69):
                    folds.append((tr, te))
            except Exception:
                folds = []

# Fallback: if splitting produced no folds, emit a single all-train / empty-test fold
if not folds:
    folds = [(np.arange(len(__xpart_src_69)), np.array([], dtype=int))]

loop_state = {
    'node_id': '69',
    'k': len(folds),
    'current': 0,
    'classColumn': _class_col,
    'stratified': bool(_y is not None),
    'shuffle': _shuffle,
    'seed': _seed,
    'folds': folds,
}
context['__loop__:69'] = loop_state
context['__flowvar__:69:maxIterations'] = len(folds)
context.pop('__xagg__:69:accum', None)
context.pop('__xagg__:69:is_complete', None)

for __fold_idx, (__tr, __te) in enumerate(folds):
    context['__flowvar__:69:currentIteration'] = int(__fold_idx)
    context['__loop__:69']['current'] = int(__fold_idx)
    train_df = __xpart_src_69.iloc[__tr].reset_index(drop=True)
    test_df  = __xpart_src_69.iloc[__te].reset_index(drop=True)
    context['69:1'] = train_df
    context['69:2'] = test_df
    ################################################################################################################################################################
    ## Equal Size Sampling # `44`
    # Node state: `EXECUTED`
    # Input: [Port 1] 69:1 from X_Partitioner #69
    # Output: [Port 1] 44:1 to 45 Gradient Boosted Trees Learner #45
    # https://hub.knime.com/knime/extensions/org.knime.features.base/latest/org.knime.base.node.preproc.equalsizesampling.EqualSizeSamplingNodeFactory
    df = context['69:1']  # input table
    out_df = df.copy()
    _class = 'Churn (Y/N)'
    _seed = 1
    if df.empty:
        out_df = df.iloc[0:0]
    else:
        groups = [g for _, g in df.groupby(_class, dropna=False, sort=False)]
        if not groups:
            out_df = df.iloc[0:0]
        else:
            min_count = min(len(g) for g in groups)
            if min_count <= 0:
                out_df = df.iloc[0:0]
            else:
                parts = [resample(g, replace=False, n_samples=min_count, random_state=_seed) for g in groups]
                out_df = pd.concat(parts, axis=0).sort_index() if parts else df.iloc[0:0]
    context['44:1'] = out_df
    ################################################################################################################################################################
    ## Gradient Boosted Trees Learner # `45`
    # Node state: `EXECUTED`
    # Input: [Port 1] 44:1 from Equal Size Sampling #44
    # Output: [Port 1] 45:1 to 49 Gradient Boosted Trees Predictor #49
    # https://hub.knime.com/knime/extensions/org.knime.features.ensembles/latest/org.knime.base.node.mine.treeensemble2.node.gradientboosting.learner.classification.GradientBoostingClassificationLearnerNodeFactory2
    df = context['44:1']  # input table
    out_df = df.copy()
    _target = 'Churn (Y/N)'
    _include = ['Address', 'Age', 'Callcard', 'Callid', 'Callwait', 'Cardmon', 'Cardten', 'Confer', 'Custcat', 'Ebill', 'Education', 'Employ', 'Equip', 'Equipmon', 'Equipten', 'Forward', 'GeFemaleder', 'Income', 'Internet', 'lnInc', 'LogLong', 'LongMon', 'LongTen', 'Marital', 'Multline', 'Pager', 'Region', 'Reside', 'Retire', 'Tenure', 'Tollfree', 'Tollmon', 'Tollten', 'Voice', 'Wireless', 'Wiremon', 'Wireten']
    feat_cols = [c for c in _include if c in df.columns]
    X = df[feat_cols].copy()
    y = df[_target].copy()
    _n_estimators = int(100)
    _learning_rate = float(0.1)
    _max_depth = 4
    _min_samples_split = int(2)
    _min_samples_leaf = int(1)
    _subsample = float(1.0)
    _max_features = None  # mapped from KNIME columnSamplingMode (None => all features)
    _seed_raw = int(1715241424300)
    _seed = int(abs(_seed_raw) % (2**32))  # coerce to [0, 2**32-1] for sklearn
    # KNIME splitCriterion='Gini' (sklearn GBT uses tree regressors internally; criterion is fixed)
    # KNIME missingValueHandling='XGBoost' (not supported in sklearn GBT; impute beforehand)
    # KNIME useAverageSplitPoints=True (not applicable in sklearn GBT)
    # KNIME useBinaryNominalSplits=True (not applicable in sklearn GBT)
    model = GradientBoostingClassifier(
        n_estimators=_n_estimators,
        learning_rate=_learning_rate,
        max_depth=_max_depth,
        min_samples_split=_min_samples_split,
        min_samples_leaf=_min_samples_leaf,
        subsample=_subsample,
        max_features=_max_features,
        random_state=_seed,
    )
    model.fit(X, y)
    fi = getattr(model, 'feature_importances_', None)
    if fi is not None:
        importances_df = pd.DataFrame({'feature': feat_cols, 'importance': fi})
        importances_df = importances_df.sort_values('importance', ascending=False, kind='mergesort').reset_index(drop=True)
    else:
        importances_df = pd.DataFrame(columns=['feature','importance'])
    summary_df = pd.DataFrame([{'n_estimators': _n_estimators, 'learning_rate': _learning_rate, 'max_depth': _max_depth, 'min_samples_split': _min_samples_split, 'min_samples_leaf': _min_samples_leaf, 'subsample': _subsample, 'max_features': _max_features, 'random_state': _seed, 'n_features': len(feat_cols), 'classes': ','.join(map(str, getattr(model, 'classes_', [])))}])
    bundle = {
        'estimator': model,
        'model': model,
        'features': list(feat_cols),
        'feature_cols': list(feat_cols),
        'target': _target,
        'classes': list(getattr(model, 'classes_', [])),
        'n_estimators': _n_estimators,
        'learning_rate': _learning_rate,
        'max_depth': _max_depth,
        'min_samples_split': _min_samples_split,
        'min_samples_leaf': _min_samples_leaf,
        'subsample': _subsample,
        'max_features': _max_features,
        'random_state': _seed,
    }
    context['45:1'] = bundle
    ################################################################################################################################################################
    ## Gradient Boosted Trees Predictor # `49`
    # Node state: `EXECUTED`
    # Input: [Port 1] 45:1 from Gradient Boosted Trees Learner #45; [Port 2] 69:2 from X_Partitioner #69
    # Output: [Port 1] 49:1 to 65 X_Aggregator #65
    # https://hub.knime.com/knime/extensions/org.knime.features.ensembles/latest/org.knime.base.node.mine.treeensemble2.node.gradientboosting.predictor.classification.GradientBoostingClassificationPredictorNodeFactory3
    model_key = '45:1'
    data_key = '69:2'
    model_obj = context[model_key]
    df = context[data_key]
    out_df = df.copy()

    # Normalize to a dict-like bundle with expected keys
    if isinstance(model_obj, dict):
        bundle = model_obj
    else:
        bundle = {'estimator': model_obj}

    est = bundle.get('estimator') or bundle.get('model')
    feat = bundle.get('features') or getattr(est, 'feature_names_in_', None)
    tgt  = bundle.get('target') or bundle.get('y_col') or bundle.get('target_name')
    classes = list(bundle.get('classes') or getattr(est, 'classes_', []))

    # Normalize features to a plain list
    if isinstance(feat, (list, tuple)):
        feat_list = list(feat)
    elif getattr(feat, 'tolist', None):
        feat_list = list(feat.tolist())
    elif feat is None:
        feat_list = []
    else:
        try:
            feat_list = list(feat)
        except Exception:
            feat_list = []

    # Feature fallback if none provided by bundle/estimator
    if not feat_list:
        if tgt and tgt in out_df.columns:
            feat_list = [c for c in out_df.columns if c != tgt]
        else:
            feat_list = list(out_df.columns)

    missing = [c for c in feat_list if c not in out_df.columns]
    if missing:
        raise KeyError(f'Missing feature columns in data: {missing}')
    X = out_df[feat_list]

    pred_col = f'Prediction ({tgt or "target"})'
    pred = est.predict(X)
    out_df[pred_col] = pd.Series(pred, index=out_df.index).astype('object')

    # Class probability columns (if supported by estimator)
    proba = None
    if hasattr(est, 'predict_proba'):
        proba = est.predict_proba(X)
        if not classes and getattr(proba, 'shape', (0, 0))[1] == 2:
            classes = ['class0', 'class1']
        _suf = '_GT'
        for j, cls in enumerate(classes):
            cname = f"P ({tgt}={cls}){_suf}"
            out_df[cname] = proba[:, j]

    # Prediction confidence (max class probability) when probabilities are available
    if proba is not None:
        conf = proba.max(axis=1)
        out_df[f"{pred_col} (confidence)"] = conf
    context['49:1'] = out_df
    ################################################################################################################################################################
    ## X_Aggregator # `65`
    # Node state: `EXECUTED`
    # Input: [Port 1] 49:1 from Gradient Boosted Trees Predictor #49
    # Output: [Port 1] 65:1 to 35 Column Appender #35; [Port 1] 65:1 to 46 Scorer #46
    # https://hub.knime.com/knime/extensions/org.knime.features.base/latest/org.knime.base.node.meta.xvalidation.AggregateOutputNodeFactory
    df = context['49:1']  # per-fold input table
    # Aggregate per-fold outputs; publish only on the final fold
    _pred_col = 'Prediction (Churn (Y/N))'
    _targ_col = 'Churn (Y/N)'
    _add_fold = False

    # Bind to an active loop state written by X-Partitioner
    _loop = None
    _loop_id = None
    # Prefer a loop that is not yet complete
    _loop_keys = [k for k in list(context.keys()) if isinstance(k, str) and k.startswith('__loop__:')]
    if _loop_keys:
        # Try to pick a loop whose accumulator is not marked complete
        _cands = []
        for _lk in _loop_keys:
            try:
                _lid = _lk.split(':', 1)[1]
            except Exception:
                _lid = None
            _lobj = context.get(_lk, None)
            _is_done = bool(context.get(f'__xagg__:{_lid}:is_complete', False)) if _lid else False
            if isinstance(_lobj, dict) and not _is_done:
                _cands.append((_lid, _lobj))
        if _cands:
            # If multiple, choose the one with the greatest 'current' (most in-progress)
            _lid, _lobj = sorted(_cands, key=lambda t: int(t[1].get('current', 0)), reverse=True)[0]
            _loop_id, _loop = _lid, _lobj
        else:
            # Fallback: last loop key (lexicographically) if all are complete
            _loop_keys.sort()
            _loop_key = _loop_keys[-1]
            _loop = context.get(_loop_key, None)
            try:
                _loop_id = _loop_key.split(':', 1)[1]
            except Exception:
                _loop_id = str(_loop_key)
    else:
        _loop = None
        _loop_id = str('65')

    _cur = int(_loop.get('current', 0)) if isinstance(_loop, dict) else 0
    _k   = int(_loop.get('k', 1))       if isinstance(_loop, dict) else 1

    # Reset accumulator at the beginning of the loop to prevent stale concatenation
    if int(_cur) == 0:
        context.pop(f'__xagg__:{_loop_id}:accum', None)
        context.pop(f'__xagg__:{_loop_id}:is_complete', None)

    # Current fold input
    cur_df = df.copy()
    if _add_fold:
        cur_df = cur_df.copy()
        cur_df['Fold'] = _cur

    # Optional: sanity-check columns (non-fatal)
    if _pred_col and _pred_col not in cur_df.columns:
        _cand = [c for c in cur_df.columns if c.startswith('Prediction (')]
        if _cand:
            _pred_col = _cand[0]
    # target column is optional for this aggregation

    # Append this fold to an accumulator under the loop id
    _acc_key = f'__xagg__:{_loop_id}:accum'
    _acc = context.get(_acc_key, None)
    if not isinstance(_acc, list):
        _acc = []
    _acc.append(cur_df)
    context[_acc_key] = _acc

    # Mark completion state and publish only on final fold
    context[f'__xagg__:{_loop_id}:is_complete'] = (len(_acc) >= max(int(_k), 1))
    if int(_cur) >= max(int(_k), 1) - 1:
        out_df = pd.concat(_acc, ignore_index=True) if _acc else cur_df.iloc[0:0].copy()
        context['65:1'] = out_df
        # (end of loop) — downstream nodes will run once, outside the loop body
```

```python
################################################################################################################################################################
## X_Partitioner # `61`
# Node state: `EXECUTED`
# Input: [Port 1] 90:1 from CSV Reader #90
# Output: [Port 2] 61:2 to 10 Decision Tree Predictor #10; [Port 1] 61:1 to 12 Equal Size Sampling #12
# https://hub.knime.com/knime/extensions/org.knime.features.base/latest/org.knime.base.node.meta.xvalidation.XValidatePartitionerFactory
__xpart_src_61 = context['90:1']  # input table (frozen for the whole CV loop)
# Plan cross-validation folds and open the loop
_leave_one_out = False
_k_req = 10
_shuffle = False
_seed = None
_stratified = True
_class_col = 'Churn (Y/N)'

n_samples = int(len(__xpart_src_61))

_y = None
if _stratified and _class_col and _class_col in __xpart_src_61.columns:
    _y = __xpart_src_61[_class_col].astype('object').where(pd.notna(__xpart_src_61[_class_col]), '__NA__')

folds = []  # list of (train_idx, test_idx)

if _leave_one_out:
    if n_samples >= 2:
        try:
            splitter = LeaveOneOut()
            for tr, te in splitter.split(__xpart_src_61, _y if _y is not None else None):
                folds.append((tr, te))
        except Exception:
            folds = []
    # else: too few samples → leave folds empty (fallback below)
else:
    # KFold / StratifiedKFold branch
    if n_samples < 2:
        folds = []  # fallback below will handle
    else:
        _k_eff = int(_k_req) if _k_req is not None else 2
        _k_eff = max(2, min(_k_eff, n_samples))
        _did_strat = False
        if _stratified and _y is not None:
            try:
                # Ensure each class has at least _k_eff members for stratification
                vc = pd.Series(_y).value_counts(dropna=False)
                if (vc.min() >= _k_eff) and (len(vc) >= 2):
                    splitter = StratifiedKFold(n_splits=_k_eff, shuffle=_shuffle, random_state=_seed if _shuffle else None)
                    for tr, te in splitter.split(__xpart_src_61, _y):
                        folds.append((tr, te))
                    _did_strat = True
            except Exception:
                folds = []
        if not _did_strat:
            try:
                splitter = KFold(n_splits=_k_eff, shuffle=_shuffle, random_state=_seed if _shuffle else None)
                for tr, te in splitter.split(__xpart_src_61):
                    folds.append((tr, te))
            except Exception:
                folds = []

# Fallback: if splitting produced no folds, emit a single all-train / empty-test fold
if not folds:
    folds = [(np.arange(len(__xpart_src_61)), np.array([], dtype=int))]

loop_state = {
    'node_id': '61',
    'k': len(folds),
    'current': 0,
    'classColumn': _class_col,
    'stratified': bool(_y is not None),
    'shuffle': _shuffle,
    'seed': _seed,
    'folds': folds,
}
context['__loop__:61'] = loop_state
context['__flowvar__:61:maxIterations'] = len(folds)
context.pop('__xagg__:61:accum', None)
context.pop('__xagg__:61:is_complete', None)

for __fold_idx, (__tr, __te) in enumerate(folds):
    context['__flowvar__:61:currentIteration'] = int(__fold_idx)
    context['__loop__:61']['current'] = int(__fold_idx)
    train_df = __xpart_src_61.iloc[__tr].reset_index(drop=True)
    test_df  = __xpart_src_61.iloc[__te].reset_index(drop=True)
    context['61:1'] = train_df
    context['61:2'] = test_df
    ################################################################################################################################################################
    ## Equal Size Sampling # `12`
    # Node state: `EXECUTED`
    # Input: [Port 1] 61:1 from X_Partitioner #61
    # Output: [Port 1] 12:1 to 9 Decision Tree Learner #9
    # https://hub.knime.com/knime/extensions/org.knime.features.base/latest/org.knime.base.node.preproc.equalsizesampling.EqualSizeSamplingNodeFactory
    df = context['61:1']  # input table
    out_df = df.copy()
    _class = 'Churn (Y/N)'
    _seed = 1
    if df.empty:
        out_df = df.iloc[0:0]
    else:
        groups = [g for _, g in df.groupby(_class, dropna=False, sort=False)]
        if not groups:
            out_df = df.iloc[0:0]
        else:
            min_count = min(len(g) for g in groups)
            if min_count <= 0:
                out_df = df.iloc[0:0]
            else:
                parts = [resample(g, replace=False, n_samples=min_count, random_state=_seed) for g in groups]
                out_df = pd.concat(parts, axis=0).sort_index() if parts else df.iloc[0:0]
    context['12:1'] = out_df
    ################################################################################################################################################################
    ## Decision Tree Learner # `9`
    # Node state: `EXECUTED`
    # Input: [Port 1] 12:1 from Equal Size Sampling #12
    # Output: [Port 1] 9:1 to 10 Decision Tree Predictor #10
    # https://hub.knime.com/knime/extensions/org.knime.features.base/latest/org.knime.base.node.mine.decisiontree2.learner2.DecisionTreeLearnerNodeFactory3
    df = context['12:1']  # input table
    out_df = df.copy()
    _target = 'Churn (Y/N)'
    num_like = df.select_dtypes(include=['number','bool','boolean','Int64','Float64']).columns.tolist()
    feat_cols = [c for c in num_like if c != _target]
    X = df[feat_cols].copy()
    y = df[_target].copy()
    _criterion = 'gini'  # mapped from KNIME splitQualityMeasure
    _min_samples_split = int(2)
    # KNIME Decision Tree Learner does not expose a random seed; k2p defaults to 1 for reproducibility.
    _seed = int(1)  # default from k2p (1 unless overridden)
    # KNIME pruningMethod='No pruning' (not directly supported by sklearn DecisionTree; consider ccp_alpha)
    # KNIME enableReducedErrorPruning=True (not supported in sklearn)
    # KNIME binaryNominalSplit=True (not applicable in sklearn)
    # KNIME maxNumNominalValues=10 (not applicable in sklearn)
    # KNIME CFG_MISSINGSTRATEGY='lastPrediction' (sklearn requires pre-imputation)
    # KNIME CFG_NOTRUECHILD='returnNullPrediction'
    model = DecisionTreeClassifier(
        criterion=_criterion,
        min_samples_split=_min_samples_split,
        random_state=_seed,
    )
    model.fit(X, y)
    fi = getattr(model, 'feature_importances_', None)
    if fi is not None:
        importances_df = pd.DataFrame({'feature': feat_cols, 'importance': fi})
        importances_df = importances_df.sort_values('importance', ascending=False, kind='mergesort').reset_index(drop=True)
    else:
        importances_df = pd.DataFrame(columns=['feature','importance'])
    summary_df = pd.DataFrame([{'criterion': _criterion, 'min_samples_split': _min_samples_split, 'n_features': len(feat_cols), 'classes': ','.join(map(str, getattr(model, 'classes_', [])))}])
    bundle = {
        'estimator': model,
        'model': model,  # alias
        'features': list(feat_cols),
        'feature_cols': list(feat_cols),
        'target': _target,
        'classes': list(getattr(model, 'classes_', [])),
        'criterion': _criterion,
        'min_samples_split': _min_samples_split,
        'random_state': _seed,
    }
    context['9:1'] = bundle
    ################################################################################################################################################################
    ## Decision Tree Predictor # `10`
    # Node state: `EXECUTED`
    # Input: [Port 1] 9:1 from Decision Tree Learner #9; [Port 2] 61:2 from X_Partitioner #61
    # Output: [Port 1] 10:1 to 66 X_Aggregator #66
    # https://hub.knime.com/knime/extensions/org.knime.features.base/latest/org.knime.base.node.mine.decisiontree2.predictor2.DecTreePredictorNodeFactory
    model_key = '9:1'
    data_key = '61:2'
    model_obj = context[model_key]
    df = context[data_key]
    out_df = df.copy()

    # Unify to a dict-like bundle with expected keys
    if isinstance(model_obj, dict):
        bundle = model_obj
    else:
        # Fallback: wrap bare estimator
        bundle = {'estimator': model_obj}

    est = bundle.get('estimator') or bundle.get('model')
    feat = bundle.get('features') or getattr(est, 'feature_names_in_', None)
    tgt  = bundle.get('target') or bundle.get('y_col') or bundle.get('target_name')
    classes = list(bundle.get('classes') or getattr(est, 'classes_', []))

    # Normalize features to a plain list
    if isinstance(feat, (list, tuple)):
        feat_list = list(feat)
    elif getattr(feat, 'tolist', None):
        feat_list = list(feat.tolist())
    elif feat is None:
        feat_list = []
    else:
        try:
            feat_list = list(feat)
        except Exception:
            feat_list = []

    # Feature fallback if none provided by bundle/estimator
    if not feat_list:
        if tgt and tgt in out_df.columns:
            feat_list = [c for c in out_df.columns if c != tgt]
        else:
            feat_list = list(out_df.columns)

    missing = [c for c in feat_list if c not in out_df.columns]
    if missing:
        raise KeyError(f'Missing feature columns in data: {missing}')
    X = out_df[feat_list]

    pred_col = f'Prediction ({tgt or "target"})'
    pred = est.predict(X)
    out_df[pred_col] = pd.Series(pred, index=out_df.index).astype('object')

    # Probability columns per class (if supported by estimator)
    if hasattr(est, 'predict_proba'):
        proba = est.predict_proba(X)
        if not classes and getattr(proba, 'shape', (0, 0))[1] == 2:
            classes = ['class0', 'class1']
        _suf = '_DT'
        for j, cls in enumerate(classes):
            cname = f"P ({tgt}={cls}){_suf}"
            out_df[cname] = proba[:, j]
    context['10:1'] = out_df
    ################################################################################################################################################################
    ## X_Aggregator # `66`
    # Node state: `EXECUTED`
    # Input: [Port 1] 10:1 from Decision Tree Predictor #10
    # Output: [Port 1] 66:1 to 11 Scorer #11; [Port 1] 66:1 to 35 Column Appender #35
    # https://hub.knime.com/knime/extensions/org.knime.features.base/latest/org.knime.base.node.meta.xvalidation.AggregateOutputNodeFactory
    df = context['10:1']  # per-fold input table
    # Aggregate per-fold outputs; publish only on the final fold
    _pred_col = 'Prediction (Churn (Y/N))'
    _targ_col = 'Churn (Y/N)'
    _add_fold = False

    # Bind to an active loop state written by X-Partitioner
    _loop = None
    _loop_id = None
    # Prefer a loop that is not yet complete
    _loop_keys = [k for k in list(context.keys()) if isinstance(k, str) and k.startswith('__loop__:')]
    if _loop_keys:
        # Try to pick a loop whose accumulator is not marked complete
        _cands = []
        for _lk in _loop_keys:
            try:
                _lid = _lk.split(':', 1)[1]
            except Exception:
                _lid = None
            _lobj = context.get(_lk, None)
            _is_done = bool(context.get(f'__xagg__:{_lid}:is_complete', False)) if _lid else False
            if isinstance(_lobj, dict) and not _is_done:
                _cands.append((_lid, _lobj))
        if _cands:
            # If multiple, choose the one with the greatest 'current' (most in-progress)
            _lid, _lobj = sorted(_cands, key=lambda t: int(t[1].get('current', 0)), reverse=True)[0]
            _loop_id, _loop = _lid, _lobj
        else:
            # Fallback: last loop key (lexicographically) if all are complete
            _loop_keys.sort()
            _loop_key = _loop_keys[-1]
            _loop = context.get(_loop_key, None)
            try:
                _loop_id = _loop_key.split(':', 1)[1]
            except Exception:
                _loop_id = str(_loop_key)
    else:
        _loop = None
        _loop_id = str('66')

    _cur = int(_loop.get('current', 0)) if isinstance(_loop, dict) else 0
    _k   = int(_loop.get('k', 1))       if isinstance(_loop, dict) else 1

    # Reset accumulator at the beginning of the loop to prevent stale concatenation
    if int(_cur) == 0:
        context.pop(f'__xagg__:{_loop_id}:accum', None)
        context.pop(f'__xagg__:{_loop_id}:is_complete', None)

    # Current fold input
    cur_df = df.copy()
    if _add_fold:
        cur_df = cur_df.copy()
        cur_df['Fold'] = _cur

    # Optional: sanity-check columns (non-fatal)
    if _pred_col and _pred_col not in cur_df.columns:
        _cand = [c for c in cur_df.columns if c.startswith('Prediction (')]
        if _cand:
            _pred_col = _cand[0]
    # target column is optional for this aggregation

    # Append this fold to an accumulator under the loop id
    _acc_key = f'__xagg__:{_loop_id}:accum'
    _acc = context.get(_acc_key, None)
    if not isinstance(_acc, list):
        _acc = []
    _acc.append(cur_df)
    context[_acc_key] = _acc

    # Mark completion state and publish only on final fold
    context[f'__xagg__:{_loop_id}:is_complete'] = (len(_acc) >= max(int(_k), 1))
    if int(_cur) >= max(int(_k), 1) - 1:
        out_df = pd.concat(_acc, ignore_index=True) if _acc else cur_df.iloc[0:0].copy()
        context['66:1'] = out_df
        # (end of loop) — downstream nodes will run once, outside the loop body
```

```python
################################################################################################################################################################
## X_Partitioner # `70`
# Node state: `EXECUTED`
# Input: [Port 1] 90:1 from CSV Reader #90
# Output: [Port 2] 70:2 to 41 SVM Predictor #41; [Port 1] 70:1 to 43 Equal Size Sampling #43
# https://hub.knime.com/knime/extensions/org.knime.features.base/latest/org.knime.base.node.meta.xvalidation.XValidatePartitionerFactory
__xpart_src_70 = context['90:1']  # input table (frozen for the whole CV loop)
# Plan cross-validation folds and open the loop
_leave_one_out = False
_k_req = 10
_shuffle = False
_seed = None
_stratified = True
_class_col = 'Churn (Y/N)'

n_samples = int(len(__xpart_src_70))

_y = None
if _stratified and _class_col and _class_col in __xpart_src_70.columns:
    _y = __xpart_src_70[_class_col].astype('object').where(pd.notna(__xpart_src_70[_class_col]), '__NA__')

folds = []  # list of (train_idx, test_idx)

if _leave_one_out:
    if n_samples >= 2:
        try:
            splitter = LeaveOneOut()
            for tr, te in splitter.split(__xpart_src_70, _y if _y is not None else None):
                folds.append((tr, te))
        except Exception:
            folds = []
    # else: too few samples → leave folds empty (fallback below)
else:
    # KFold / StratifiedKFold branch
    if n_samples < 2:
        folds = []  # fallback below will handle
    else:
        _k_eff = int(_k_req) if _k_req is not None else 2
        _k_eff = max(2, min(_k_eff, n_samples))
        _did_strat = False
        if _stratified and _y is not None:
            try:
                # Ensure each class has at least _k_eff members for stratification
                vc = pd.Series(_y).value_counts(dropna=False)
                if (vc.min() >= _k_eff) and (len(vc) >= 2):
                    splitter = StratifiedKFold(n_splits=_k_eff, shuffle=_shuffle, random_state=_seed if _shuffle else None)
                    for tr, te in splitter.split(__xpart_src_70, _y):
                        folds.append((tr, te))
                    _did_strat = True
            except Exception:
                folds = []
        if not _did_strat:
            try:
                splitter = KFold(n_splits=_k_eff, shuffle=_shuffle, random_state=_seed if _shuffle else None)
                for tr, te in splitter.split(__xpart_src_70):
                    folds.append((tr, te))
            except Exception:
                folds = []

# Fallback: if splitting produced no folds, emit a single all-train / empty-test fold
if not folds:
    folds = [(np.arange(len(__xpart_src_70)), np.array([], dtype=int))]

loop_state = {
    'node_id': '70',
    'k': len(folds),
    'current': 0,
    'classColumn': _class_col,
    'stratified': bool(_y is not None),
    'shuffle': _shuffle,
    'seed': _seed,
    'folds': folds,
}
context['__loop__:70'] = loop_state
context['__flowvar__:70:maxIterations'] = len(folds)
context.pop('__xagg__:70:accum', None)
context.pop('__xagg__:70:is_complete', None)

for __fold_idx, (__tr, __te) in enumerate(folds):
    context['__flowvar__:70:currentIteration'] = int(__fold_idx)
    context['__loop__:70']['current'] = int(__fold_idx)
    train_df = __xpart_src_70.iloc[__tr].reset_index(drop=True)
    test_df  = __xpart_src_70.iloc[__te].reset_index(drop=True)
    context['70:1'] = train_df
    context['70:2'] = test_df
    ################################################################################################################################################################
    ## Equal Size Sampling # `43`
    # Node state: `EXECUTED`
    # Input: [Port 1] 70:1 from X_Partitioner #70
    # Output: [Port 1] 43:1 to 39 SVM Learner #39
    # https://hub.knime.com/knime/extensions/org.knime.features.base/latest/org.knime.base.node.preproc.equalsizesampling.EqualSizeSamplingNodeFactory
    df = context['70:1']  # input table
    out_df = df.copy()
    _class = 'Churn (Y/N)'
    _seed = 1
    if df.empty:
        out_df = df.iloc[0:0]
    else:
        groups = [g for _, g in df.groupby(_class, dropna=False, sort=False)]
        if not groups:
            out_df = df.iloc[0:0]
        else:
            min_count = min(len(g) for g in groups)
            if min_count <= 0:
                out_df = df.iloc[0:0]
            else:
                parts = [resample(g, replace=False, n_samples=min_count, random_state=_seed) for g in groups]
                out_df = pd.concat(parts, axis=0).sort_index() if parts else df.iloc[0:0]
    context['43:1'] = out_df
    ################################################################################################################################################################
    ## SVM Learner # `39`
    # Node state: `EXECUTED`
    # Input: [Port 1] 43:1 from Equal Size Sampling #43
    # Output: [Port 1] 39:1 to 41 SVM Predictor #41
    # https://hub.knime.com/knime/extensions/org.knime.features.base/latest/org.knime.base.node.mine.svm.learner.SVMLearnerNodeFactory2
    df = context['43:1']  # input table
    out_df = df.copy()
    _target = 'Churn (Y/N)'
    if _target not in df.columns:
        raise KeyError(f"SVM Learner: target column not found: {_target!r}")
    num_like = df.select_dtypes(include=['number','bool','boolean','Int64','Float64']).columns.tolist()
    feat_cols = [c for c in num_like if c != _target]
    if not feat_cols:
        raise ValueError('SVM Learner: no numeric/bool feature columns found (after excluding target)')
    X = df[feat_cols].copy()
    y = df[_target].copy()
    _mask = y.notna()
    if not bool(_mask.all()):
        X = X.loc[_mask]
        y = y.loc[_mask]
    if X.shape[0] == 0:
        raise ValueError('SVM Learner: no training rows after dropping missing target values')
    _kernel = 'poly'
    _C = float(1.0)
    _gamma = float(1.0)
    _degree = int(1)
    _coef0 = float(1.0)
    _probability = bool(True)
    _class_weight = None
    _seed = int(1)  # used for probability calibration
    model = SVC(
        kernel=_kernel,
        C=_C,
        gamma=_gamma,
        degree=_degree,
        coef0=_coef0,
        probability=_probability,
        class_weight=_class_weight,
        random_state=_seed,
    )
    model.fit(X, y)
    coef_df = pd.DataFrame(columns=['feature','coef'])
    if hasattr(model, 'coef_') and getattr(model, 'coef_', None) is not None:
        try:
            coeffs = model.coef_
            if getattr(coeffs, 'ndim', 1) == 1:
                coef_df = pd.DataFrame({'feature': feat_cols, 'coef': coeffs})
            else:
                tmp = []
                for i, row in enumerate(coeffs):
                    tmp.append(pd.DataFrame({'feature': feat_cols, 'coef': row, 'pair': i}))
                coef_df = pd.concat(tmp, ignore_index=True)
            coef_df = coef_df.sort_values(by='coef', key=lambda s: s.abs(), ascending=False, kind='mergesort').reset_index(drop=True)
        except Exception:
            coef_df = pd.DataFrame(columns=['feature','coef'])
    summary_df = pd.DataFrame([{'kernel': _kernel, 'C': _C, 'gamma': _gamma, 'degree': _degree, 'coef0': _coef0, 'n_features': len(feat_cols), 'classes': ','.join(map(str, getattr(model, 'classes_', [])))}])
    bundle = {
        'estimator': model,
        'model': model,  # alias
        'features': list(feat_cols),
        'feature_cols': list(feat_cols),
        'target': _target,
        'classes': list(getattr(model, 'classes_', [])),
        'kernel': _kernel,
        'C': _C,
        'gamma': _gamma,
        'degree': _degree,
        'coef0': _coef0,
    }
    context['39:1'] = bundle
    ################################################################################################################################################################
    ## SVM Predictor # `41`
    # Node state: `EXECUTED`
    # Input: [Port 1] 39:1 from SVM Learner #39; [Port 2] 70:2 from X_Partitioner #70
    # Output: [Port 1] 41:1 to 67 X_Aggregator #67
    # https://hub.knime.com/knime/extensions/org.knime.features.base/latest/org.knime.base.node.mine.svm.predictor2.SVMPredictorNodeFactory
    model_key = '39:1'
    data_key = '70:2'
    model_obj = context[model_key]
    df = context[data_key]
    out_df = df.copy()

    # Normalize to a dict-like bundle with expected keys
    if isinstance(model_obj, dict):
        bundle = model_obj
    else:
        # Fallback: wrap bare estimator
        bundle = {'estimator': model_obj}

    est = bundle.get('estimator') or bundle.get('model')
    if est is None:
        raise ValueError('SVM Predictor: missing estimator in model bundle')
    feat = bundle.get('features') or getattr(est, 'feature_names_in_', None)
    tgt  = bundle.get('target') or bundle.get('y_col') or bundle.get('target_name')
    classes = list(bundle.get('classes') or getattr(est, 'classes_', []))

    # Normalize features to a plain list
    if isinstance(feat, (list, tuple)):
        feat_list = list(feat)
    elif getattr(feat, 'tolist', None):
        feat_list = list(feat.tolist())
    elif feat is None:
        feat_list = []
    else:
        try:
            feat_list = list(feat)
        except Exception:
            feat_list = []

    # Feature fallback if none provided by bundle/estimator
    if len(feat_list) == 0:
        if tgt and tgt in out_df.columns:
            feat_list = [c for c in out_df.columns if c != tgt]
        else:
            feat_list = list(out_df.columns)

    missing = [c for c in feat_list if c not in out_df.columns]
    if missing:
        raise KeyError(f'SVM Predictor: missing feature columns in data: {missing}')
    X = out_df[feat_list]

    pred_col = f'Prediction ({tgt or "target"})'
    _suf = '_SV'

    if X.shape[0] == 0:
        # Create empty prediction/probability columns and publish.
        out_df[pred_col] = pd.Series(index=out_df.index, dtype='object')
        if bool(True):
            _cls = classes or list(getattr(est, 'classes_', []))
            for cls in _cls:
                cname = f"P ({tgt}={cls}){_suf}"
                out_df[cname] = pd.Series(index=out_df.index, dtype='float64')
    else:
        # Normal scoring path
        pred = est.predict(X)
        out_df[pred_col] = pd.Series(pred, index=out_df.index).astype('object')
        if bool(True) and hasattr(est, 'predict_proba'):
            try:
                proba = est.predict_proba(X)
            except Exception:
                proba = None
            if proba is not None:
                # Align column order with estimator.classes_ when available
                _cls_order = list(getattr(est, 'classes_', [])) or list(classes)
                for j, cls in enumerate(_cls_order):
                    cname = f"P ({tgt}={cls}){_suf}"
                    out_df[cname] = proba[:, j]
    context['41:1'] = out_df
    ################################################################################################################################################################
    ## X_Aggregator # `67`
    # Node state: `EXECUTED`
    # Input: [Port 1] 41:1 from SVM Predictor #41
    # Output: [Port 1] 67:1 to 35 Column Appender #35; [Port 1] 67:1 to 40 Scorer #40
    # https://hub.knime.com/knime/extensions/org.knime.features.base/latest/org.knime.base.node.meta.xvalidation.AggregateOutputNodeFactory
    df = context['41:1']  # per-fold input table
    # Aggregate per-fold outputs; publish only on the final fold
    _pred_col = 'Prediction (Churn (Y/N))'
    _targ_col = 'Churn (Y/N)'
    _add_fold = False

    # Bind to an active loop state written by X-Partitioner
    _loop = None
    _loop_id = None
    # Prefer a loop that is not yet complete
    _loop_keys = [k for k in list(context.keys()) if isinstance(k, str) and k.startswith('__loop__:')]
    if _loop_keys:
        # Try to pick a loop whose accumulator is not marked complete
        _cands = []
        for _lk in _loop_keys:
            try:
                _lid = _lk.split(':', 1)[1]
            except Exception:
                _lid = None
            _lobj = context.get(_lk, None)
            _is_done = bool(context.get(f'__xagg__:{_lid}:is_complete', False)) if _lid else False
            if isinstance(_lobj, dict) and not _is_done:
                _cands.append((_lid, _lobj))
        if _cands:
            # If multiple, choose the one with the greatest 'current' (most in-progress)
            _lid, _lobj = sorted(_cands, key=lambda t: int(t[1].get('current', 0)), reverse=True)[0]
            _loop_id, _loop = _lid, _lobj
        else:
            # Fallback: last loop key (lexicographically) if all are complete
            _loop_keys.sort()
            _loop_key = _loop_keys[-1]
            _loop = context.get(_loop_key, None)
            try:
                _loop_id = _loop_key.split(':', 1)[1]
            except Exception:
                _loop_id = str(_loop_key)
    else:
        _loop = None
        _loop_id = str('67')

    _cur = int(_loop.get('current', 0)) if isinstance(_loop, dict) else 0
    _k   = int(_loop.get('k', 1))       if isinstance(_loop, dict) else 1

    # Reset accumulator at the beginning of the loop to prevent stale concatenation
    if int(_cur) == 0:
        context.pop(f'__xagg__:{_loop_id}:accum', None)
        context.pop(f'__xagg__:{_loop_id}:is_complete', None)

    # Current fold input
    cur_df = df.copy()
    if _add_fold:
        cur_df = cur_df.copy()
        cur_df['Fold'] = _cur

    # Optional: sanity-check columns (non-fatal)
    if _pred_col and _pred_col not in cur_df.columns:
        _cand = [c for c in cur_df.columns if c.startswith('Prediction (')]
        if _cand:
            _pred_col = _cand[0]
    # target column is optional for this aggregation

    # Append this fold to an accumulator under the loop id
    _acc_key = f'__xagg__:{_loop_id}:accum'
    _acc = context.get(_acc_key, None)
    if not isinstance(_acc, list):
        _acc = []
    _acc.append(cur_df)
    context[_acc_key] = _acc

    # Mark completion state and publish only on final fold
    context[f'__xagg__:{_loop_id}:is_complete'] = (len(_acc) >= max(int(_k), 1))
    if int(_cur) >= max(int(_k), 1) - 1:
        out_df = pd.concat(_acc, ignore_index=True) if _acc else cur_df.iloc[0:0].copy()
        context['67:1'] = out_df
        # (end of loop) — downstream nodes will run once, outside the loop body
```

```python
################################################################################################################################################################
## Scorer # `40`
# Node state: `EXECUTED`
# Input: [Port 1] 67:1 from X_Aggregator #67
# Output: [Port 1] 40:1 to 88 CSV Writer #88
# https://hub.knime.com/knime/extensions/org.knime.features.base/latest/org.knime.base.node.mine.scorer.accuracy.AccuracyScorer2NodeFactory
df = context['67:1']  # input table with truth & prediction
out_df = df.copy()  # not strictly required, but keeps pattern consistent
_truth_col = 'Churn (Y/N)'
_pred_col  = 'Prediction (Churn (Y/N))'

# Validate required columns exist
missing = []
if _truth_col not in df.columns: missing.append(_truth_col)
if _pred_col not in df.columns:  missing.append(_pred_col)
if missing:
    raise KeyError(f'Scorer: missing required column(s): {missing}')

pair = df[[_truth_col, _pred_col]].copy()
pair = pair.dropna(subset=[_truth_col, _pred_col])  # KNIME: ignore.missing.values = true

y_true = pair[_truth_col].astype('object')
y_pred = pair[_pred_col].astype('object')

# Metrics
acc = float(accuracy_score(y_true, y_pred)) if len(pair) else float('nan')
err = (1.0 - acc) if pd.notna(acc) else float('nan')
correct = int((y_true == y_pred).sum())
false   = int((y_true != y_pred).sum())
try:
    kappa = float(cohen_kappa_score(y_true, y_pred)) if len(pair) else float('nan')
except Exception:
    kappa = float('nan')

# Confusion matrix with a stable label list (union from both columns, order of appearance)
labels = pd.unique(pd.concat([y_true, y_pred], ignore_index=True))
if len(labels) == 0:
    cm_df = pd.DataFrame(columns=pd.Index([], name='Predicted'), index=pd.Index([], name='Actual'))
else:
    cm = confusion_matrix(y_true, y_pred, labels=labels)
    cm_df = pd.DataFrame(cm, index=pd.Index(labels, name='Actual'), columns=pd.Index(labels, name='Predicted'))

summary_rows = [
    {'Metric': 'Accuracy',       'Value': acc},
    {'Metric': 'Error',          'Value': err},
    {'Metric': '#Correct',       'Value': correct},
    {'Metric': '#False',         'Value': false},
    {'Metric': "Cohen's kappa", 'Value': kappa},
]
summary_df = pd.DataFrame(summary_rows)

# Outputs:
# Port 1 → confusion matrix, Port 2 → summary metrics
cm_out = cm_df
stats_out = summary_df
print(cm_df)
print(summary_df)
context['40:1'] = cm_out
context['40:2'] = stats_out
```

## Column Appender \# `35`
Node state: `EXECUTED`  
Input: [Port 5] 62:1 from X_Aggregator #62; [Port 3] 63:1 from X_Aggregator #63; [Port 4] 64:1 from X_Aggregator #64; [Port 2] 65:1 from X_Aggregator #65; [Port 1] 66:1 from X_Aggregator #66; [Port 6] 67:1 from X_Aggregator #67  
Output: [Port 1] 35:1 to 36 ROC Curve #36  


```python
################################################################################################################################################################
## Column Appender # `35`
# Node state: `EXECUTED`
# Input: [Port 5] 62:1 from X_Aggregator #62; [Port 3] 63:1 from X_Aggregator #63; [Port 4] 64:1 from X_Aggregator #64; [Port 2] 65:1 from X_Aggregator #65; [Port 1] 66:1 from X_Aggregator #66; [Port 6] 67:1 from X_Aggregator #67
# Output: [Port 1] 35:1 to 36 ROC Curve #36
# https://hub.knime.com/knime/extensions/org.knime.features.base/latest/org.knime.base.node.preproc.columnappend2.ColumnAppender2NodeFactory
left_df = context['66:1']  # left table
right_dfs = []
right_dfs.append(context['65:1'])
right_dfs.append(context['63:1'])
right_dfs.append(context['64:1'])
right_dfs.append(context['62:1'])
right_dfs.append(context['67:1'])
out_df = left_df.copy()
_mode = 'IDENTICAL'

def _unique_with_suffix(base: str, existing: set, suffix: str) -> str:
    if base not in existing:
        return base
    new = f"{base}{suffix}"
    # If even the suffixed name exists, keep appending the same suffix
    while new in existing:
        new = f"{new}{suffix}"
    return new

# Append each right table in order (1-based index for suffix)
for k, rdf in enumerate(right_dfs, start=1):
    suffix = f" (#{k})"
    # Build a collision-aware rename map guaranteeing uniqueness
    existing = set(out_df.columns)
    rename_map = {}
    for c in rdf.columns:
        if c in existing or c in rename_map.values():
            newc = _unique_with_suffix(c, existing | set(rename_map.values()), suffix)
            rename_map[c] = newc
        else:
            # no collision; keep original
            rename_map[c] = c
    right_safe = rdf.rename(columns=rename_map) if rename_map else rdf

    if _mode == 'IDENTICAL':
        # Align by index; if different, left-align by reindexing right to out_df.index
        if out_df.index.equals(right_safe.index):
            out_df = out_df.join(right_safe, how='left')
        else:
            out_df = out_df.join(right_safe.reindex(out_df.index), how='left')
    else:
        # Positional fallback for unknown modes
        out_df = pd.concat([out_df.reset_index(drop=True),
                           right_safe.reset_index(drop=True)], axis=1)
context['35:1'] = out_df
```

## ROC Curve \# `36`
Node state: `EXECUTED`  
Input: [Port 1] 35:1 from Column Appender #35  


```python
################################################################################################################################################################
## ROC Curve # `36`
# Node state: `EXECUTED`
# Input: [Port 1] 35:1 from Column Appender #35
# https://hub.knime.com/knime/extensions/org.knime.features.base/latest/org.knime.base.views.node.roccurve.ROCCurveNodeFactory
df = context['35:1']  # input table with y and probabilities
_truth_col = 'Churn (Y/N) (#2)'
_pos_label = 'Y'
_proba_cols = ['P (Churn (Y/N)=Y)_DT', 'P (Churn (Y/N)=Y)_RF', 'P (Churn (Y/N)=Y)_GT', 'P (Churn (Y/N)=Y)_NN', 'P (Churn (Y/N)=Y)_LR', 'P (Churn (Y/N)=Y)_SV']
_title = 'ROC Curve'
_x_label = 'False positive rate (1 - specificity)'
_y_label = 'True positive rate (sensitivity)'
_width_in = 800 / 100.0
_height_in = 600 / 100.0
_img_fmt = 'PNG'
_line_width = int(2) if 2 else 2

# Validate inputs
if _truth_col is None or _truth_col not in df.columns:
    raise KeyError(f"ROC: ground-truth column not found: {_truth_col!r}")
if _pos_label is None:
    raise KeyError("ROC: positive class value is not configured in settings.xml")
if not _proba_cols:
    raise KeyError('ROC: no probability columns configured in settings.xml.')

# Keep only configured probability columns that are present in the table
proba_cols = [c for c in _proba_cols if c in df.columns]
if not proba_cols:
    raise KeyError(f"ROC: none of the configured probability columns are present: {_proba_cols}")
missing = [c for c in _proba_cols if c not in df.columns]
if missing:
    print(f"[ROC] Warning: missing probability columns ignored: {missing}")

# Prepare y and iterate curves
y_true = df[_truth_col].astype('object')
y_bin = (y_true == _pos_label).astype(int)

rows = []
for col in proba_cols:
    # Coerce to [0,1] numeric probabilities
    p = pd.to_numeric(df[col], errors='coerce').fillna(0.0).astype(float).clip(0.0, 1.0)
    fpr, tpr, thr = roc_curve(y_bin, p)
    auc_val = float(auc(fpr, tpr))
    tmp = pd.DataFrame({'model': col, 'fpr': fpr, 'tpr': tpr, 'threshold': thr})
    tmp['auc'] = auc_val
    rows.append(tmp)

points_df = pd.concat(rows, ignore_index=True) if rows else pd.DataFrame(columns=['model','fpr','tpr','threshold','auc'])

# Plot
plt.figure(figsize=(_width_in, _height_in), dpi=100)
for m, grp in points_df.groupby('model', sort=False):
    auc_val = float(grp['auc'].iloc[0]) if not grp.empty else float('nan')
    plt.plot(grp['fpr'].values, grp['tpr'].values, linewidth=_line_width, label=f"{m} (AUC={auc_val:.3f})")
plt.plot([0,1], [0,1], linestyle='--', linewidth=1)
plt.title(_title)
plt.xlabel(_x_label)
plt.ylabel(_y_label)
plt.xlim(0, 1); plt.ylim(0, 1)
plt.legend(loc='lower right')

# Robust output directory (cwd may be unavailable in some notebook kernels)
try:
    out_dir = Path.cwd()
    out_dir.mkdir(parents=True, exist_ok=True)
except Exception:
    out_dir = Path(tempfile.gettempdir()) / 'knime2py_roc'
    out_dir.mkdir(parents=True, exist_ok=True)
ext = 'svg' if str(_img_fmt).upper() == 'SVG' else 'png'
img_path = out_dir / ('roc_36.' + ext)
csv_path = out_dir / ('roc_table_36.csv')
plt.savefig(img_path, bbox_inches='tight')
plt.close()
points_df.to_csv(csv_path, index=False)
print(f'[ROC] Wrote image to: {img_path}')
print(f'[ROC] Wrote points to: {csv_path}')
```
